```python
import numpy as np
```

```python
# create an array using numpy
arr = np.array([1,2,3,4,5])
print(arr)
```

```python
# 2d array create
arr2 = np.array([
    [1,2,3,4],
    [4,5,6,7],
    [7,8,9,10]
])
print(arr2.shape) #3X4
```

```python
#np.arange(start,stop,step)
arr3 = np.arange(0,10,2) # stop is not include
print(arr3)
```

```python
arr4 = np.arange(5) # this is considered as stop and start is assumed 0
print(arr4)
```

```python
arr5 = np.ones((3,4)) # array of 3*4 with all ones
# ek extra bracket
print(arr5)
```

```python
arr6 = np.eye(3) # identity matrix od 3*3 only diag ele are 1
print(arr6)
```

```python
arr = np.array([[1,2,3],[4,5,6]])

print(arr.shape)#(2,3)
print(arr.ndim)#2 as 2d
print(arr.size) #6
print(arr.dtype)#int64
print(arr.itemsize)#8
```

```python
#numpy vectorised operations
arr1 = np.array([1,2,3,4,5])
arr2 = np.array([10,20,30,40,50])

#element wise addition
print(arr1 + arr2)
```

```python
numbers = [1, 2, 3, 4, 5]

result = []

for x in numbers:
    result.append(x * 2)

print(result)
```

```python
numbers = np.array([1, 2, 3, 4, 5])

result = numbers * 2

print(result)
```

```python
#universal function : a function that can work on evey element of a numpy array automatically

arr = np.array([1,4,9,16])
print(np.sqrt(arr))#like vectorisation

print(np.exp(arr)) #e^x

print(np.sin(arr))
```

```python
#NumPy Indexing and Slicing

arr = np.array([10,20,30,40,50])

print(arr[0]) #0 based indexing

#negative indexing
print(arr[-1])#50


#slicing : getting multiple elements from array
#arr[start:stop:step]
```

```python
a = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

#indexing
print(a[1, 1])

#slicing
print(a[:2 , :2]) # pehle konsi row , phir konsi col
```

```python
#modify elements
a = np.array([10, 20, 30, 40, 50])

a[2] = 100
print(a)

a[1:4] = 99
print(a)

a[1:4] = [200, 300, 400]
print(a)
```

```python
#normalisation : mean of 0 and std dev as 1

data = np.array([1,2,3,4,5])
mean = np.mean(data)
std_dev = np.std(data)

normalized_data = (data - mean) / std_dev

print(normalized_data)
```

```python
data = np.arange(0,11)

#mean
mean = np.mean(data)
print(mean)

#median
median = np.median(data)
print(median)

#std dev
std_dev = np.std(data)
print(std_dev)

#variance
variance = np.var(data)
print(variance)
```

```python
# logical operations

a = np.array([10, 20, 30, 40, 50])
print(a>30)
```

```python
# filtering

print(a[a>30])
```

ADVANCED CONCEPTS IN NUMPY


```python
# axis in numpy : it tells in which direction to perform operation

a = np.array([
    [1,2,3],
    [4,5,6]
])
#axis = 0 -> column wise

print(np.sum(a))
print(np.sum(a,axis=0)) #[5,7,9]
```

```python
#aggregate function : reducing many values into a single value
a = np.array([1,2,3,4,5])

print(np.sum(a))
print(np.mean(a))
```

```python
# sort
a = np.array([1,2,5,6,3,2])
print(np.sort(a)) # sorted
print(a) # not changes original

a.sort()
print(a) # changes original
```

```python
# concatenating
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

c = np.concatenate((a, b))

print(c) #[1 2 3 4 5 6]
```

```python
# stacking
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

#vertical stacking
print(np.vstack((a, b)))
#horizontal stacking
print(np.hstack((a, b)))
```

```python
#copy vs view

a = np.array([1, 2, 3, 4])

b = a.view()

b[0] = 100

print(a) #[100 2 3 4] # changing b also changed a


a = np.array([1, 2, 3, 4])

b = a.copy()

b[0] = 100

print(a) # [1,2,3,4]
```

```python
x =np.random.randint(1,10) # this generates a random integer from 1 to 9
print(x)

y = np.random.randint(1,10,5) # 5 numbers
print(y)
```

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

# element wise multiplication
print(A*B)

#matrix multiplication
print(A@B)
```

```python
# linear algebra
A = np.array([
    [1, 2],
    [3, 4]
])
print(np.linalg.det(A))       # determinant
print(np.linalg.inv(A))       # inverse
print(np.linalg.eig(A))       # eigenvalues/eigenvectors
```

```python
# broadcasting

# array + single number
a = np.array([1,2,3])
print(a+10)

#[1  2  3]
#[10 10 10]
#---------
#[11 12 13]
```

```python
#2d array + 1darray
a = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

b = np.array([10, 20, 30])

print(a+b)
```

```python
a = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

b = np.array([
    [10],
    [20]
])

print(a + b)
```
